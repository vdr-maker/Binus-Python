```python
# ===== Program File Handling =====

def buat_file():
    try:
        nama_file = input("Masukkan Nama File: ")

        nama = input("Masukkan Namamu: ")
        nim = input("Masukkan NIM kamu: ")
        angkatan = input("Masukkan tahun angkatanmu: ")

        file = open(nama_file + ".txt", "w")
        file.write("Nama: " + nama + "\n")
        file.write("NIM: " + nim + "\n")
        file.write("Angkatan: " + angkatan + "\n")

        print("\nFile Berhasil Dibuat")

    except Exception:
        print("\nTerjadi kesalahan! File tidak dapat dibuat.")

    finally:
        print("Proses pembuatan file selesai.")


def baca_file():
    file = None

    try:
        nama_file = input("Masukkan Nama File: ")

        file = open(nama_file + ".txt", "r")
        isi_file = file.read()

        print("\n")
        print(isi_file)

    except FileNotFoundError:
        print("\nFile tidak ditemukan.")

    except Exception:
        print("\nTerjadi kesalahan! File tidak dapat dibuka.")

    finally:
        if file is not None:
            file.close()

        print("Proses membaca file selesai.")


def tambah_text():
    file = None

    try:
        nama_file = input("Masukkan Nama File: ")

        nama_sahabat = input("Masukkan Nama Sahabatmu: ")
        catatan = input("Masukkan Catatan Tambahan kamu: ")

        file = open(nama_file + ".txt", "a")
        file.write("\nNama Sahabat: " + nama_sahabat + "\n")
        file.write("Catatan: " + catatan + "\n")

        print("\nData Berhasil Ditambahkan")

    except FileNotFoundError:
        print("\nFile tidak ditemukan.")

    except Exception:
        print("\nTerjadi kesalahan! Data tidak dapat ditambahkan.")

    finally:
        if file is not None:
            file.close()

        print("Proses penambahan data selesai.")


def program_file_handling():
    pilihan = ""

    while pilihan != "4":
        print("\n===== Program File Handling =====")
        print("1. Membuat dan Menulis File")
        print("2. Membaca File")
        print("3. Menambahkan Text Pada File")
        print("4. Keluar Dari Program")

        pilihan = input("\nMasukkan Pilihan Berupa Angka (1/2/3/4): ")

        if pilihan == "1":
            buat_file()

        elif pilihan == "2":
            baca_file()

        elif pilihan == "3":
            tambah_text()

        elif pilihan == "4":
            print("Terima Kasih Sudah Menggunakan Program Saya")

        else:
            print("Pilihan tidak tersedia. Silakan pilih angka 1-4.")


# Menjalankan program
program_file_handling()
```
