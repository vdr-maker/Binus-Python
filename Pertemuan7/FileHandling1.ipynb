```python
def tulis_biodata():
    nama = input("Nama: ")
    usia = input("Usia: ")
    alamat = input("Alamat: ")
    email = input("Email: ")

    with open("Biodata.txt", "w") as file:
        file.write(f"Nama: {nama}\n")
        file.write(f"Usia: {usia}\n")
        file.write(f"Alamat: {alamat}\n")
        file.write(f"Email: {email}\n")


def baca_biodata():
    with open("Biodata.txt", "r") as file:
        isi = file.read()

    print("\nIsi Biodata.txt:")
    print(isi)


# Program utama
tulis_biodata()
baca_biodata()
```
