```python
import pandas as pd

# Membaca file CSV
df = pd.read_csv("data_negara.csv", sep=';')

print("===== Data Negara ====")
print(df)

# Rata-rata
rata_rata = df["Populasi_Juta"].mean()

# Deviasi standar
deviasi_standar = df["Populasi_Juta"].std()

print("\n===== Hasil Perhitungan ====")
print("Rata-rata :", rata_rata)
print("Deviasi Standar :", deviasi_standar)
```
